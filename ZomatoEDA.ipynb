**Average Rating by Location**

* Koramangala 5th Block has the highest average rating among the top locations (~3.9).
* Indiranagar and Jayanagar also perform well.

**Insight:**

**2. Location Insights**

* Koramangala 5th Block and Indiranagar show stronger average ratings.

* BTM offers high competition with relatively lower pricing and average ratings → value segment is crowded.

**3.Cuisine Insights**
* North Indian is heavily saturated (highest volume).
* South Indian and Biryani also face high competition with average ratings.
* Continental, Desserts, and Cafe deliver the best ratings.
* Continental restaurants have the highest engagement and command premium pricing.

**4.Feature Impact (Online Order & Book Table)**

Online Ordering :
* Small positive effect on rating
* Clear positive effect on number of votes (higher engagement)

Table Booking :
* Strong positive association with rating (4.11 vs 3.57)
* Extremely high engagement (almost 9× more votes)

**Key Recommendations for a New Restaurant Owner**
1. Avoid pure high-volume, low-differentiation formats (basic Quick Bites / generic North Indian) unless you have a strong unique angle.
2. Consider Cafe or Dessert-focused concepts — better ratings with manageable competition.
3. Premium dine-in (Casual Dining with table booking) performs well on both rating and engagement, but requires higher investment.
4. Offering table booking appears to be a strong quality signal in the Bengaluru market.
5. Location choice matters more than just density — Koramangala 5th Block and Indiranagar show better quality signals than pure volume leaders like BTM.

## Advanced EDA

### Underserved Cuisine–Location Pairs

**We want to identify white-space opportunities:**
>Cuisines that perform well overall (good average rating),but have very few or zero restaurants in a particular location.
* These are potential gaps a new restaurant owner could fill.

**Logic We Should Use**

1. We need two levels of information:
* Overall Cuisine Strength
* Average rating of the cuisine across Bengaluru
* Total number of restaurants serving that cuisine

2. Local Presence
* How many restaurants of that cuisine exist in each specific location
* Average rating of that cuisine in that location (if any)

```python
# Top 15 locations by restaurant count
top15_locations = (df['LOCATION_CLEAN'].value_counts().head(15).index.tolist() )
print(top15_locations)
```

```python
# Filter data for only Top 15 locations
df_top15=df[df['LOCATION_CLEAN'].isin(top15_locations)]
```

```python
# Create Location + Primary Cuisine combination
#local_presence
local_presence = (df_top15.groupby(['LOCATION_CLEAN', 'PRIMARY_CUISINE']).agg(
        local_count = ('NAME', 'count'),
        local_avg_rate = ('RATE', 'mean'),
        local_avg_vote = ('VOTES', 'mean')).reset_index() )

print(local_presence.head(20))
print("\nShape:", local_presence.shape)
```

```python
#distribution of local_count
print(local_presence['local_count'].describe())
print(local_presence['local_avg_rate'].describe())
print(local_presence['local_avg_vote'].describe())

print("\nValue counts of local_count:")
print(local_presence['local_count'].value_counts().sort_index().head(20))
```

**Distribution Insights**
* Total combinations (Top 15 locations × Primary Cuisine): 629
* Median restaurants per combination: 3
* 186 combinations have only 1 restaurant
* 80 combinations have only 2 restaurants
* 49 combinations have only 3 restaurants

```python
#Overall cuisine performance
cuisine_overall = (df.groupby('PRIMARY_CUISINE').agg(
    overall_count=("NAME","count"),
    overall_avg_rate=("RATE","mean"),
    overall_avg_vote=("VOTES","mean")).reset_index())
cuisine_overall.head()
```

```python
local_presence.head(5)
```

```python
underserved =local_presence.merge(cuisine_overall,on="PRIMARY_CUISINE",how="left")
underserved.head(5)
```

> We'll create opportunity score to make more balanced score (o_avg_rate*2) + (o_avg_vote/100) - local_count
* **overall_avg_rate * 2**   = Rewards cuisines that are highly rated overall,Higher quality cuisines are more attractive,Strong positive
* **overall_avg_vote / 100** = Rewards cuisines that have good customer engagement,High votes = proven popularity,Moderate positive
* **-(local_count * 1.5)**    = Penalizes combinations that already have many restaurants,More existing restaurants = less opportunity (more competition),Strong negative

```python

underserved['opportunity_score'] = ((underserved['overall_avg_rate'] * 2) + (underserved['overall_avg_vote'] / 100) - 
    (underserved['local_count'] * 1.5)
)
```

```python
underserved['opportunity_score'].describe()
```

```python
# normalise the score 0-100 for better understanding
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler(feature_range=(0, 100))
underserved['opportunity_score_normalized'] = scaler.fit_transform( underserved[['opportunity_score']] )
```

```python
print(underserved.head(5))
```

```python
underserved_final = underserved[(underserved['local_count'] <= 4) & (underserved['overall_avg_rate'] >= 3.7) & \
    (underserved['opportunity_score_normalized'] >= 85) ].sort_values(
    ['opportunity_score_normalized', 'local_count'],
    ascending=[False, True])

underserved_final.drop(columns=['opportunity_score'])
```

```python
underserved_final.groupby(['LOCATION_CLEAN'])['opportunity_score_normalized'].mean()
```

```python
underserved_final.groupby(['LOCATION_CLEAN','PRIMARY_CUISINE'])['opportunity_score_normalized'].mean()
```

```python
top15_opp = underserved_final.head(15).copy()

# Create a combined label for better readability
top15_opp['label'] = top15_opp['LOCATION_CLEAN'] + " - " + top15_opp['PRIMARY_CUISINE']

plt.figure(figsize=(14, 8))
ax=sns.barplot(data=top15_opp,y='label',x='opportunity_score_normalized',hue='label',palette='coolwarm')
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)
plt.title('Top 15 Underserved Cuisine-Location Opportunities', fontsize=14)
plt.xlabel('Opportunity Score (0-100)')
plt.ylabel('Location - Cuisine')
plt.tight_layout()
plt.show()
```

```python
# Average Opportunity Score by Location (from the underserved list)
location_opp = (underserved_final.groupby('LOCATION_CLEAN').agg(avg_opportunity_score = ('opportunity_score_normalized', 'mean'),\
        opportunity_count = ('PRIMARY_CUISINE', 'count')).reset_index().sort_values('avg_opportunity_score', ascending=False) )

plt.figure(figsize=(12, 6))
ax=sns.barplot(data=location_opp,x='LOCATION_CLEAN',y='avg_opportunity_score',hue='avg_opportunity_score',palette='magma')
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)
plt.xticks(rotation=45, ha='right')
plt.title('Average Opportunity Score by Location (Underserved Cuisines)', fontsize=14)
plt.ylabel('Average Opportunity Score (0-100)')
plt.xlabel('Location')
plt.show()
```

```python
plt.figure(figsize=(12, 6))
ax=sns.barplot(data=location_opp.sort_values('opportunity_count', ascending=False),x='LOCATION_CLEAN',\
    y='opportunity_count', hue='opportunity_count',palette='crest')
for container in ax.containers:
    ax.bar_label(container, fmt='%.2f', padding=3)
plt.xticks(rotation=45, ha='right')
plt.title('Number of Underserved Cuisine Opportunities by Location')
plt.ylabel('Number of Underserved Cuisines')
plt.xlabel('Location')
plt.tight_layout()
plt.show()
```

- **High-Potential Cuisines:** Premium and international cuisines such as **European, Continental, and American** are under-represented despite strong customer ratings and engagement.
